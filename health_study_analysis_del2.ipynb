### Blood Pressure and Age

I created a scatter plot to look at the relationship between age and blood pressure.

The graph shows how blood pressure varies among people of different ages. It helps me explore whether age may be related to blood pressure.

```python
analyzer = HealthAnalyzer(data)

coefficients = analyzer.multiple_regression()

print("Regression coefficients:")
print("Intercept:", coefficients[0])
print("Age:", coefficients[1])
print("Weight:", coefficients[2])
```

### Multiple Linear Regression

I used NumPy to study how age and weight are related to blood pressure.

The model uses both age and weight to estimate blood pressure.

The results show that both variables have a positive relationship with blood pressure. However, this does not mean they are the only factors that affect it.

```python
print("Multiple regression completed successfully!")
print("Number of coefficients:", len(coefficients))
```
