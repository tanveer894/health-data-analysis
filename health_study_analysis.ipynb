### 95% Confidence Interval for Mean Systolic Blood Pressure

```python
# Calculate mean and standard error
mean_bp = df["systolic_bp"].mean()
std_bp = df["systolic_bp"].std()
n = len(df)

standard_error = std_bp / np.sqrt(n)

# Calculate 95% confidence interval
lower_limit = mean_bp - 1.96 * standard_error
upper_limit = mean_bp + 1.96 * standard_error

print("Mean systolic blood pressure:", mean_bp)
print("95% Confidence Interval:", lower_limit, "to", upper_limit)
```

### Interpretation of the Confidence Interval

The average systolic blood pressure in the dataset is 149.18 mmHg.

The 95% confidence interval is from 148.29 to 150.07 mmHg.

This means that we are 95% confident that the true population mean systolic blood pressure lies within this interval.

```python
# Separate blood pressure values for smokers and non-smokers
smokers_bp = df[df["smoker"] == "Yes"]["systolic_bp"]
non_smokers_bp = df[df["smoker"] == "No"]["systolic_bp"]

# Calculate mean blood pressure
smokers_mean = smokers_bp.mean()
non_smokers_mean = non_smokers_bp.mean()

# Perform independent samples t-test
from scipy.stats import ttest_ind

t_statistic, p_value = ttest_ind(
    smokers_bp,
    non_smokers_bp,
    alternative="greater"
)

print("Smokers mean blood pressure:", smokers_mean)
print("Non-smokers mean blood pressure:", non_smokers_mean)
print("T-statistic:", t_statistic)
print("P-value:", p_value)
```

```python
# Separate blood pressure values
smokers_bp = df[df["smoker"] == "Yes"]["systolic_bp"]
non_smokers_bp = df[df["smoker"] == "No"]["systolic_bp"]

# Calculate group means
smokers_mean = smokers_bp.mean()
non_smokers_mean = non_smokers_bp.mean()

# Calculate sample sizes
n1 = len(smokers_bp)
n2 = len(non_smokers_bp)

# Calculate standard deviations
s1 = smokers_bp.std()
s2 = non_smokers_bp.std()

# Calculate t-statistic
standard_error = np.sqrt((s1**2 / n1) + (s2**2 / n2))

t_statistic = (smokers_mean - non_smokers_mean) / standard_error

# Display results
print("Smokers mean blood pressure:", smokers_mean)
print("Non-smokers mean blood pressure:", non_smokers_mean)
print("T-statistic:", t_statistic)
```

```python
# Calculate degrees of freedom (Welch's t-test)
df1 = (s1**2 / n1 + s2**2 / n2)**2

df2 = ((s1**2 / n1)**2 / (n1 - 1)) + ((s2**2 / n2)**2 / (n2 - 1))

degrees_of_freedom = df1 / df2

# Calculate one-sided p-value without SciPy
from math import sqrt
from scipy.special import stdtr

p_value = stdtr(degrees_of_freedom, -t_statistic)

print("Degrees of freedom:", degrees_of_freedom)
print("P-value:", p_value)
```

### Interpretation of Hypothesis Testing

The average systolic blood pressure among smokers is 149.52 mmHg, while the average among non-smokers is 149.05 mmHg.

The p-value is 0.3264, which is greater than the significance level of 0.05.

Therefore, we fail to reject the null hypothesis. There is not enough statistical evidence to conclude that smokers have a higher mean blood pressure than non-smokers.
